```python
import pandas as pd
df = pd.read_csv("movies.csv")

df
```

```python
df.iloc[0]
```

```python
df = df.drop(["original_title","budget","homepage","original_language","popularity","production_countries","production_companies","release_date","revenue","runtime","spoken_languages","status","vote_average","vote_count"],axis = 1)
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset = ['overview']) 
df['tagline'] = df['tagline'].fillna(' ')   #this will fill all the free tagline with the empty string as we will merge the tagline with other data
```

```python
df.isnull().sum()  #the output represent 0 missing value
```

```python
import ast #this is to convert string to list
```

```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
df
```

```python
df['genres'] = df['genres'].apply(convert)
```

```python
df['keywords'] = df['keywords'].apply(convert)
```

```python
df['overview'] = df['overview'].apply(lambda x:x.split())
df['tagline'] = df['tagline'].apply(lambda x:x.split())
```

```python
df['genres'] = df['genres'].apply(lambda x:[i.replace(" ","")for i in x])
df['overview'] = df['overview'].apply(lambda x:[i.replace(" ","")for i in x])
df['tagline'] = df['tagline'].apply(lambda x:[i.replace(" ","")for i in x])
df['keywords'] = df['keywords'].apply(lambda x:[i.replace(" ","")for i in x])
```

```python
df
```

```python
df['tags'] = df['overview'] + df['genres'] + df['keywords'] + df['tagline']
```

```python
new_df = df[['id','title','tags']]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
#steming : to short the words to their root words ["loving", "loved","lovely"] all these word will get converted to love

import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 5000,stop_words = 'english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
print(cv.get_feature_names_out())
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
#this is the actual funtion that recommend the top 5 similar movies

def recommend(movie): 
    movie_index = new_df[new_df['title'] == movie].index[0]     #this fetch the movie index of the title
    distance = similarity[movie_index]                         #this tell the similarities array of that particular movie with others
    
     #the enumerate fun will allot the index to the cosine similarity then it will be converted into a list and then it will be sorted in reverse manner
     #based on the similarity score so on top we will get the movie that we are searching for as its similarity score with itself will be the highest
    #then we can print from the 2nd movie to the 6th to get the most similar movie
    
    movie_list = sorted(list(enumerate(distance)),reverse = True,key = lambda x:x[1]) 
    
    for i in movie_list[1:6]:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
